```python
year = int(input("Enter a year: "))

if year % 4 == 0:          # First check divisibility by 4
    if year % 100 == 0:    # Then check divisibility by 100
        if year % 400 == 0:  # Finally check divisibility by 400
            print(f"{year} is a leap year.")
        else:
            print(f"{year} is not a leap year.")
    else:
        print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
```
